```python
import time
from random import randint, random
from copy import deepcopy
from rat_maze import rat_maze_game
from utils import get_manhattan_distance

class genetic_algo():
    moveLimit = 30
    populationSize = 1000
    # less chance at mutation
    mutationRate = 0.3 
    
    #  for already optimized chromosome define an operation called no move 
    #i.e if the chromosome succeeded in the most optimal moves say 40, then remaining  60 moves should be no moves
     # no move, up, down, left, right
        
    genesName = ['N', 'U', 'D', 'L', 'R']
    genesValue = [0, 1, 2, 3, 4]
    MAX_GENERATIONS = 4

    def __init__(self):
        #initilize the population and generation 
        self.curretPopulation = []
        self.generation = 1


    def best_generation(self, start_time):
        best = deepcopy(sorted(
            self.curretPopulation, key=lambda chromosome: self.fitness_function(chromosome)
        )[:3])

        for i in range(len(best)):
            for j in range(self.moveLimit):
                best[i][j] = self. genesName[best[i][j]]

        print('Generation {} Time taken - {}s'.format(self.generation, time.monotonic() - start_time))
        for item in best:
            score = self.fitness_function(item)
            print(''.join(item), '(Score: {})'.format(score[0]))

    
    def base_population(self):
        self.curretPopulation = [
            [
                randint(self.genesValue[0], self.genesValue[-1]) for y in range(self.moveLimit)
            ] for x in range(self. populationSize )
        ]

    def fitness_function(self, chromosome):
        game = rat_maze_game()

        # play the game
        for move in chromosome:
            if move == 1 or move == 'U':
                game.up()
            elif move == 2 or move == 'D':
                game.down()
            elif move == 3 or move == 'L':
                game.left()
            elif move == 4 or move == 'R':
                game.right()

        # distance
        distance = get_manhattan_distance(game.x, game.y, game.x_food, game.y_food)

        # add penalty if there is a wall between current position and goal
        extra_penalty = game.isWall()

        # minimize this score! tuple(score, moves)
        return (distance * extra_penalty + game.penalty, game.position)

    # crossover applies on a subset of the population that are fittest
    def crossover(self):
        newPopulation = []

        # sort population by fitness value then number of moves
        self.curretPopulation.sort(key=lambda chromosome: self.fitness_function(chromosome))

        # consider 25% fittest population
        subset = self.curretPopulation[:self. populationSize  // 4]
        subsetSize = len(subset)

        for i in range(self.populationSize):
            Parent1 = subset[randint(0, subsetSize - 1)]
            Parent2 = subset[randint(0, subsetSize - 1)]

            # select a random crossover point in the genespace
            crossoverPoint = randint(0, self.moveLimit - 1)

            newPopulation.append(
                self.mutation(Parent1[:crossoverPoint] + Parent2[crossoverPoint:])
            )
        
        self.curretPopulation = newPopulation

    def mutation(self, chromosome):
        if random() < self.mutationRate:
            chromosome[randint(0, self.moveLimit - 1)] = randint(self.genesValue[0], self.genesValue[-1])
        
        return chromosome

    
    def run(self):
        t = time.monotonic()

        self.base_population()
        self.best_generation(t)

        for i in range(1, self.MAX_GENERATIONS):
            t = time.monotonic()
            self.generation += 1

            self.crossover()
            self.best_generation(t)

if __name__ == '__main__':
    a = genetic_algo()
    a.run()
```
